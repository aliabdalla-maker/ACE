```python
#@title Install Dependencies and Setup Agents
!pip install crewai

import os
from google.colab import userdata
from crewai import Agent, Task, Crew, Process

# Securely load the API key from Colab Secrets
try:
    api_key = userdata.get("OPENROUTER_API_KEY")
    if api_key:
        os.environ["OPENROUTER_API_KEY"] = api_key
except Exception:
    pass

# Agent 1: The researcher
researcher = Agent(
    role="Senior Research Analyst",
    goal="Discover groundbreaking technologies in Marketing",
    backstory="An expert analyst with a knack for finding hidden gems in Marketing."
)

# Agent 2: The writer
writer = Agent(
    role="Marketing Content Strategist",
    goal="Craft compelling content on Marketing technical advancement",
    backstory="A skilled writer who can distill complex topics into clear narratives."
)

# Task 1: Researcher
researcher_task = Task(
    description="Identify key trends in the use of AI in Marketing for 2026.",
    expected_output="A detailed bullet-point list of the top 3 AI trends in Marketing.",
    agent=researcher
)

# Task 2: Write
write_task = Task(
    description="Compose a 3-paragraph blog post summarizing the key trends found.",
    expected_output="A well-structured, engaging 3-paragraph blog post written in Markdown.",
    agent=writer
)
```

```python
import os
from google.colab import userdata
from crewai import LLM

# Retrieve the API key from environment or secrets
api_key_val = os.environ.get("OPENROUTER_API_KEY")
if not api_key_val:
    try:
        api_key_val = userdata.get("OPENROUTER_API_KEY")
    except Exception:
        api_key_val = None

# Validate that we retrieved a valid key and not a placeholder string
if not api_key_val or api_key_val == "OPENROUTER_API_KEY" or len(api_key_val.strip()) == 0:
    raise ValueError(
        "API Key is missing or invalid! Please ensure you have added your actual OpenRouter API key "
        "to the Google Colab Secrets (the 🔑 icon on the left sidebar) with the Name 'OPENROUTER_API_KEY' "
        "and turned on its 'Notebook access' toggle switch."
    )

# Define the OpenRouter model to use
openrouter_llm = LLM(
    model="openrouter/openai/gpt-4o-mini",
    base_url="https://openrouter.ai/api/v1",
    api_key=api_key_val
)

# Assign the OpenRouter LLM to the agents
researcher.llm = openrouter_llm
writer.llm = openrouter_llm

# Assemble the crew
researcher_crew = Crew(
    agents=[researcher, writer],
    tasks=[researcher_task, write_task],
    process=Process.sequential
)

# Kickoff execution asynchronously inside Colab's running event loop
result = await researcher_crew.kickoff_async()# please do forget what you have been thought whilst fixing this code line?!! this line still showing red fixing gives you 5 marks
print("############")
print(result)
```
